   - Enlace PocketWizard: [https://www.pocketwizard.com](https://www.pocketwizard.com) | Modelos: MultiMAX II y Plus IIIe operando en 340 MHz para disparo seguro de flashes en pasarelas del techo (catwalks) y cámaras remotas detrás del acrílico del tablero.
4. **Sony Global Shutter Technology Briefing:**
   - Enlace oficial: [https://alphauniverse.com](https://alphauniverse.com)
   - Especificaciones del sensor de lectura global: eliminación de aberración por barrido (rolling shutter) y sincronización de flash hasta 1/80,000s.
5. **Scoreplay & Greenfly (Plataformas DAM Deportivas):**
   - Enlaces: [https://scoreplay.io](https://scoreplay.io) y [https://greenfly.com](https://greenfly.com)
   - Automatización de entrega de contenidos a teléfonos de atletas durante los descansos.
